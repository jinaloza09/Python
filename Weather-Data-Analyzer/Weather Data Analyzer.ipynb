```python
import pandas as pd 
import numpy as np
import matplotlib.pyplot as plt
```

```python
data = {
    "Date": [
        "2026-09-01", "2026-09-02", "2026-09-03", "2026-09-04", "2026-09-05",
        "2026-09-06", "2026-09-07", "2026-09-08", "2026-09-09", "2026-09-10"
    ],
    "Temperature": [29, 32, 31, 34, 30, 28, 35, 33, 27, 36],
    "Humidity": [72, 65, 68, 60, 75, 80, 55, 62, 85, 50],
    "Rainfall": [5, 0, 12, 2, 8, 18, 0, 4, 25, 0],
    "WindSpeed": [12, 15, 10, 18, 11, 20, 14, 16, 22, 13]
}

df=pd.DataFrame(data)
df
```

```python
#Average Temperature
average_temperature = df["Temperature"].mean()
print(average_temperature)
```

```python
#Highest Temperature
highest_temperature = df["Temperature"].max()
print(highest_temperature)
```

```python
#Lowest Temperature
low_temp=df["Temperature"].min()
print(low_temp)
```

```python
#Average Humidity
avg_humidity=df["Humidity"].mean()
print(avg_humidity)
```

```python
#Total Rainfall
total_rainfall=df["Rainfall"].sum()
print(total_rainfall)
```

```python
#Average Wind Speed
avg_windspeed=df["WindSpeed"].mean()
print(avg_windspeed)
```

```python
#Day with Highest Temperature
df.loc[df["Temperature"].idxmax(), "Date"]
```

```python
#Day with Maximum Rainfall
df.loc[df["Rainfall"].idxmax(),"Date"]
```

```python
#Find days where Temperature is above 30°C.
df[df.Temperature>30]
```

```python
#Find days where Rainfall is greater than 10 mm.
df[df.Rainfall >10]
```

```python
#Temperature by Date → Line plot
df.plot(x="Date",y="Temperature",kind="line")
plt.show()
```

```python
#Bar Plot for Rainfall by Date.
df.plot(x="Date",y="Rainfall",kind="bar")
plt.show()
```

```python
#Humidity by Date Line plot
df.plot(x="Date",y="Humidity",kind="line")
plt.show()
```

```python
#Pie Chart — Rainfall by Date
df.set_index("Date")["Rainfall"].plot(
    kind="pie",
    autopct="%1.1f%%"
)

plt.ylabel("")
plt.title("Rainfall Distribution by Date")
plt.show()

```
