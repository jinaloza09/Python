```python
#odd and even
def odd_even(num):
    if num%2==0:
        print(num,"is an even number")
    else:
        print(num,"is an odd number")
```

```python
odd_even(18)
```

```python
#factorial number
def factnum(n):
    fact =1
    
    for i in range(1,n+1):
        fact =fact*i
    return fact
factnum(5)
```

```python
#some of natural number
def sum_natural (n):
    num=0
    for i in range(1,n+1): 
        num=num+i #0+1=1   1+2=3   3+3=6   6+4=10
    return num
sum_natural(4)
```
