```python
#pin=int(input("Enter your pin"))
balance=10000
# Transaction History
transactions = []

#PIN Authentication
correct_pin = 1234
attempts = 0
max_attempts = 3

while attempts < max_attempts:

    pin = int(input("Enter your PIN: "))

    if pin == correct_pin:
        print("PIN is correct")
        print("Access granted!")
        break
    else:
        attempts = attempts + 1
        print("Invalid PIN")
        print("Attempts remaining:", max_attempts - attempts)
else:
    print("Account locked!")
    print("Too many incorrect attempts.")
    
#check balance
def check_balance():
    global balance
    print("Your Current Balance is :",balance)
    
#deposit money
def deposit_money(deposit):
    global balance 
    balance=deposit+balance
    return balance
    
#withdraw money    
def withdraw_money(withdrawamt):
    global balance

    if withdrawamt<=balance:
        balance=balance-withdrawamt
        print(" Withdrawal Successfull")
        return balance
    else:
        print("Insuffient balance")
        
#transfer money
def transfer_money(amt):
    global balance
    if amt <= 0 :
        print("Invalid transfer amount")
    elif amt <= balance:
        balance = balance - amt
        print("Transfer Successful")
        print("Your Balance is:", balance)
    else:
        print("Insufficient Balance")
        
#changing pin
def change_pin():

    global correct_pin
    old_pin = int(input("Enter your old PIN: "))
    if old_pin == correct_pin:
        new_pin=int(input("Enter your New PIN:"))
        confirm_pin=int(input("Enter your Confirm new PIN:"))
        if new_pin == confirm_pin:
            correct_pin = new_pin
            print("your PIN has been Change")
        else:
            print("New pin does not match")
    else:
        print("Incorrect PIN \n please try again")
    

while True:
    print("\n =======🏧 ATM MANAGEMENT SYSTEM ======")
    print("1. Check Balance")
    print("2. Deposit Money")
    print("3. Withdraw Money")
    print("4. Transfer Money")
    print("5. Change PIN")
    print("6. Exit")
    print("\n ========================================")

    Choice=int(input("Enter your Choice:"))

    if Choice == 1:
        check_balance()
    elif Choice == 2:
        deposit=int(input("Enter your Deposit Amount"))
        balance=deposit_money(deposit)
        print("Deposit Successfull")
        print("Your Balance is:",balance)
    elif Choice == 3:
        withdrawamt=int(input("Enter your withdrawal amount :"))
        balance=withdraw_money(withdrawamt)
        print("Your Balance is:",balance)
    elif Choice == 4:
        Account_no=int(input("Enter Your Account NO:"))
        amt=int(input("Enter  Amount:"))
        amt=transfer_money(amt)
        print("Your Balance is:",balance)
    elif Choice == 5:
        change_pin()

    elif Choice == 6:
        print("Thank you for using our ATM!")
        break
    else:
        print("Invalid Transcation")
```
