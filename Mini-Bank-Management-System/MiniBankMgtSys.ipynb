```python
name=input("Enter your name")
account_no=int(input("Enter your account number"))
balance=0
# Deposit money
def deposit_money(deposit):
    global balance
    balance = balance + deposit
    return balance

# Deposit ₹500
balance = deposit_money(500)

print("Your balance is:", balance)
#withdraw money
def withdraw_money(withdrawamt):
    global balance

    if withdrawamt <= balance:
        balance = balance - withdrawamt
        print("Withdrawal successful")
        return balance
    else:
        print("Insufficient balance")
        return balance
# Check balance
def check_balance():
    print("Your current balance is:", balance)
# Account details
def account_details():
    print("\n----- Account Details -----")
    print("Name:", name)
    print("Account Number:", account_no)
    print("Balance:", balance)
# Main menu
while True:

    print("\n===== MINI BANK =====")
    print("1. Deposit Money")
    print("2. Withdraw Money")
    print("3. Check Balance")
    print("4. Account Details")
    print("5. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:

        deposit = int(input("Enter deposit amount: "))
        balance = deposit_money(deposit)
        print("Deposit successful")
        print("Your balance is:", balance)

    elif choice == 2:

        withdrawamt = int(input("Enter withdrawal amount: "))
        balance = withdraw_money(withdrawamt)
        print("Your balance is:", balance)

    elif choice == 3:

        check_balance()

    elif choice == 4:

        account_details()

    elif choice == 5:

        print("Thank you for using Mini Bank!")
        break

    else:

        print("Invalid choice. Please try again.")
```
